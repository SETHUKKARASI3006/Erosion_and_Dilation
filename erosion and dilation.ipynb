```python
# Import the necessary packages
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def load_img():
    image = np.zeros((600,600))
    text = 'ARASI'
    font = cv2.FONT_HERSHEY_SIMPLEX
    cv2.putText(image,text,(100,350),font,5,(255),25)
    return image
```

```python
def display(img,title):
    plt.title(title)
    plt.imshow(img, cmap = 'gray')
    plt.axis('on')
    plt.show()
```

```python
# Create the text using cv2.putText
image = load_img()
```

```python
display(image, 'TEXT IMAGE')
```

```python
# Create the structuring element
kernel = np.ones((3,3))
```

```python
# Erode the image
ero1 = cv2.erode(image,kernel,iterations = 5)
```

```python
display(ero1,'Eroded image')
```

```python
# Dilate the image
dilated = cv2.dilate(image, kernel, iterations =3)
```

```python
display(dilated, 'Dilated Image')
```
